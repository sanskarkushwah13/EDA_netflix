```python
# Additional EDA techniques: missing-value profiling, country/genre trends, duration patterns, and cast concentration

# 1) Missing value proportions
missing_pct = (df.isna().mean() * 100).sort_values(ascending=False)
print('Missing value percentages by column:')
print(missing_pct)

missing_pct.plot(kind='bar', figsize=(10, 5), color='darkorange')
plt.title('Missing Values Percentage by Column')
plt.ylabel('Missing %')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2) Country distribution (explode multi-country values)
country_series = df['country'].fillna('Unknown').str.split(',').explode().str.strip()
country_counts = country_series.value_counts().head(10)
print('\nTop 10 countries by title count:')
print(country_counts)

country_counts.plot(kind='barh', figsize=(10, 6), color='seagreen')
plt.title('Top 10 Countries by Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.tight_layout()
plt.show()

# 3) Genre analysis using listed_in column
genre_series = df['listed_in'].fillna('Unknown').str.split(',').explode().str.strip()
genre_counts = genre_series.value_counts().head(10)
print('\nTop 10 genres:')
print(genre_counts)

genre_counts.plot(kind='bar', figsize=(10, 5), color='purple')
plt.title('Top 10 Netflix Genres')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 4) Type-by-year trend using pivot table
year_type = df.pivot_table(index='release_year', columns='type', aggfunc='size', fill_value=0)
print('\nYear-wise Netflix titles by type (last 10 years):')
print(year_type.tail(10))

year_type.plot(kind='line', figsize=(12, 5), marker='o')
plt.title('Netflix Release Trends by Type')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.legend(title='Type')
plt.tight_layout()
plt.show()

# 5) Duration analysis for movies
movies = df[df['type'] == 'Movie'].copy()
movies['duration_minutes'] = movies['duration'].str.extract(r'(\d+)').astype(float)
print('\nMovie duration summary:')
print(movies['duration_minutes'].describe())

movies.groupby('release_year')['duration_minutes'].mean().plot(kind='line', figsize=(12, 5), marker='o', color='crimson')
plt.title('Average Movie Duration by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Average Duration (minutes)')
plt.tight_layout()
plt.show()

# 6) Cast concentration analysis
cast_series = df['cast'].fillna('').str.split(',').explode().str.strip()
cast_series = cast_series[cast_series != '']
cast_counts = cast_series.value_counts().head(10)
print('\nTop 10 actors by number of titles:')
print(cast_counts)

cast_counts.plot(kind='barh', figsize=(10, 6), color='darkblue')
plt.title('Top 10 Actors by Netflix Appearances')
plt.xlabel('Number of Titles')
plt.ylabel('Actor')
plt.tight_layout()
plt.show()

# 7) Cross-tab of type by rating
rating_type = pd.crosstab(df['type'], df['rating'])
print('\nRating distribution by type:')
print(rating_type.head())

sns.heatmap(rating_type, annot=False, cmap='coolwarm')
plt.title('Rating Distribution by Content Type')
plt.tight_layout()
plt.show()

# 8) Quick insights summary
print('\nQuick insights:')
print(f"Most frequent country: {country_counts.index[0]} with {country_counts.iloc[0]} titles")
print(f"Most frequent genre: {genre_counts.index[0]} with {genre_counts.iloc[0]} titles")
print(f"Top cast member: {cast_counts.index[0]} with {cast_counts.iloc[0]} titles")
print(f"Highest average movie duration year: {movies.groupby('release_year')['duration_minutes'].mean().idxmax()}")
```
